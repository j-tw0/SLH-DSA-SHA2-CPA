# Key-generation WOTS_PRF path: chained compression CPA

This notebook corresponds to the paper's **key-generation-derived WOTS_PRF**
experiment: 3 independent SK.seed values per SHA2 parameter set, one complete
top-level XMSS WOTS_PRF keygen address sequence per seed, using cumulative-HW
8-bit split CPA on updated A. It uses **recovered-only chaining** to evaluate
full SK.seed recovery. The keygen capture schema is validated without copying
ChipWhisperer firmware/capture code into this CPA repository. This is not a
continuous whole-key-generation waveform.


```python
from pathlib import Path
import os
import sys

REPO_ROOT = Path.cwd().resolve()
if not (REPO_ROOT / "src").is_dir():
    REPO_ROOT = REPO_ROOT.parent
if not (REPO_ROOT / "src").is_dir():
    raise FileNotFoundError("Start Jupyter from the repository root or notebooks directory")
sys.path.insert(0, str(REPO_ROOT / "src"))

import keygen_wots_cpa as runner
```

## Configuration

Check the dataset path and keep `RUN_CPA = False` until ready.


```python
BASE_FOLDER = Path(os.environ.get("SLH_CPA_KEYGEN_BASE", "./data/keygen"))
RESULTS_DIR = Path(os.environ.get("SLH_CPA_KEYGEN_RESULTS", "./results/keygen"))
PARAM_SETS = ("128f", "128s", "192f", "192s", "256f", "256s")
SEED_IDS = "0,1,2"
COMPILER_OPTS = ("Os",)
BACKEND = "auto"
RUN_CPA = False
```

## Data inventory and optional CPA execution

Dry runs inspect local measured datasets; they do not generate new measurements.


```python
def argv(output):
    return ["--base", str(BASE_FOLDER), "--output", str(output),
            "--params", ",".join(PARAM_SETS), "--opts", ",".join(COMPILER_OPTS),
            "--seed-ids", SEED_IDS, "--protocol", "chained",
            "--max-traces", "0", "--backend", BACKEND,
            "--beam", "4", "--no-plots"]

if not BASE_FOLDER.is_dir():
    print("Missing local capture folder:", BASE_FOLDER)
else:
    print("Preflight:", runner.main(argv(RESULTS_DIR / "_preview") + ["--dry-run"]))
    if RUN_CPA:
        print("CPA:", runner.main(argv(RESULTS_DIR)))
    else:
        print("Preview only. Set RUN_CPA=True after checking the dataset and parameters.")
```
